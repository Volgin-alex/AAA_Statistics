## Подводка

Рассмортим на практических метриках например: `Средний чек`  пользователя это не тривиальное распределение. Но оцнивая средний чек нас интересует для Генеральной совокупности мат ожидание = $\mu$

$\xi \sim ?$

Чтобы проверить мат. ож. нужно придумать статистику и определить её распределение.
Для этого нам нужна **ЦПТ:** $\sum\limits_{i=1}^{n}\xi_i$, то при n->бесконечности сумма наблюдений будет **распределена по нормальному закону** $Normal(n\mu, n\sigma^2)$

## Что такое норм распределение?

$f_{Norm(\mu,\sigma^2)(x)} = \frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{1(x-\mu)^2}{2\sigma^2}}$**Плотность нормального распределения**
